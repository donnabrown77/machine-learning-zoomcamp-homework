```python
# calculate the number of missing values
missing_values = df.isnull().any().sum()
print(missing_values)
```

```python
# What's the maximum fuel efficiency of cars from Asia?
max_fuel_efficiency = df[df.origin == 'Asia'].fuel_efficiency_mpg.max()
print(max_fuel_efficiency)
```

```python
# what is the median value of horsepower?
# Find the median value of horsepower column in the dataset.
median_value = df.horsepower.median()
print(median_value)
# Next, calculate the most frequent value of the same horsepower column.
most_frequent = df.horsepower.mode()[0]
print(most_frequent)
# Use fillna method to fill the missing values in horsepower column with the most frequent value from the previous step.
df.horsepower = df.horsepower.fillna(most_frequent)
# Now, calculate the median value of horsepower once again.
median_value = df.horsepower.median()
print(median_value)
# Yes the value increased
```

```python
# Q7 Sum of weights
# Select all the cars from Asia
cars_from_asia = df[
    df.origin == 'Asia'
]

# Select only columns vehicle_weight and model_year
cars_from_asia = cars_from_asia[[
    'vehicle_weight',
    'model_year'
]]  
# Select the first 7 values
first_seven = cars_from_asia.head(7)
# Get the underlying NumPy array. Let's call it X.
X = first_seven.to_numpy()
# Compute matrix-matrix multiplication between the transpose of X and X. To get the transpose,
# use X.T. Let's call the result XTX.
XTX = X.T.dot(X)
# Invert XTX. Let's call the result XTX_inv.
XTX_inv = np.linalg.inv(XTX)
# Create an array y with the following values: [1100, 1300, 800, 900, 1000, 1100, 1200].
y = pd.Series([1100, 1300, 800, 900, 1000, 1100, 1200])
# Multiply the inverse of XTX with the transpose of X, and then multiply the result by y. Call the result w.
w = XTX_inv.dot(X.T).dot(y)
# What's the sum of all the elements of the result?
result_sum = w.sum()
print(result_sum)
```
